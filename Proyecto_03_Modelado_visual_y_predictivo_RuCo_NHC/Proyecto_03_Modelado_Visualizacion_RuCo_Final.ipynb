## Proyecto_03 Modelado visual y predictivo. Intercambio isotópico H/D en N-heterociclos catalizado por nanopartículas bimetálicas Ru/Co·NHC (NHC = IMes)

**Introducción:**

El marcaje isotópico de compuestos orgánicos presenta un largo historial en química médica, con ejemplos reportados desde inicios del siglo XX. Debido a la facilidad de detección de átomos de deuterio y tritio mediante distintas técnicas analíticas, la sustitución de átomos de hidrógeno por sus isótopos más pesados permite señalizar las moléculas de interés sin modificar su estructura química ni propiedades físicas. Particularmente, en el caso de los átomos de deuterio, su presencia en un principio activo dado puede modificar la distribución, el metabolismo y la excreción del fármaco. Por este motivo, goza de gran interés en lo que respecta al desarrollo de fármacos. Además, el marcaje isotópico con deuterio puede producir los denominados efectos isotópicos cinéticos, por lo que constituye una técnica común en el estudio de mecanismos de reacciones químicas.   

Entre los distintos métodos existentes para producir moléculas marcadas con deuterio (D) destaca el denominado *Hydrogen Isotopic Exchange* (HIE), que permite el intercambio directo entre átomos de hidrógeno y deuterio (H/D) utilizando directamente deuterio gaseoso. Las reacciones HIE permiten incorporar deuterio en la molécula objetivo sin necesidad de modificar toda la ruta sintética, aunque como contraparte, puede ser complejo controlar la selectividad del proceso. 

En este notebook, se va a utilizar python para estudiar una serie de reacciones de intercambio isotópico H/D utilizando los datos publicados en [Dalton Trans. (2025) 54 (46): 17250–17258](https://doi.org/10.1039/d5dt01969a). En total, el paper reporta como catalizadores un total de 4 nanomateriales diferentes. Estos nanomateriales están basados en sistemas de nanopartículas de rutenio y/o cobalto sintetizadas a partir del proceso sintético conocido como método organometálico. Este método consiste en la descomposición en condiciones suaves de un precursor organometálico que contiene al metal de interés. Si a dicha descomposición le acompaña la presencia de un agente estabilizante, como puede ser un ligando orgánico, es posible producir nanopartículas de manera muy controlada, en lo que respecta a tamaño, forma y estado superficial. En este caso, se ha utilizado un carbeno, IMes, como ligando estabilizante, y distintas relaciones molares Ru:Co.

Por otro lado, en las condiciones de reacción, que implican presión de deuterio gaseoso, la reacción de intercambio compite con la propia reducción de la molécula, pudiéndose eliminar insaturaciones de la misma. Este último caso constituye un proceso químico no deseado.
```python
#Caso de la molécula 2-methylquinoline, s3. Cabe destacar que con el catalizador Ru·IMes se produjo una reducción superior al 99%, por lo que no tiene sentido incluir valores en las columnas A-F, que hacen referencia a intercambio isotópico H/D. Se incluye None en las posiciones correspondientes A-F para Ru·IMes. 
print("Sustrato s3, 2-methylquinoline")
display(Image(filename="s3_2-methylquinoline.png")) #Mostramos la estructura y posiciones potencialmente marcables con deuterio
tabla_s3= {
    "Catalizador": ["Ru·IMes", "Ru90Co10·IMes", "Ru50Co50·IMes", "Ru10Co90·IMes"],
    "%Red": [99, 85, 75, 0],
    "A": [None, 91, 92, 0],
    "B": [None, 0, 15, 0],
    "C": [None, 11, 13, 0],
    "E": [None, 16, 58, 0],
    "F": [None, 0, 2, 0],
}
df_s3 = pd.DataFrame(tabla_s3) #Convertimos el diccionario en un dataframe de pandas
print(df_s3)
```

```python
#Caso de la molécula imidazole, s4
print("Sustrato s4, imidazole")
display(Image(filename="s4_imidazole.png")) #Mostramos la estructura y posiciones potencialmente marcables con deuterio
tabla_s4= {
    "Catalizador": ["Ru·IMes", "Ru90Co10·IMes", "Ru50Co50·IMes", "Ru10Co90·IMes"],
    "%Red": [0, 0, 0, 0],
    "A": [98, 98, 98, 0],
    "B": [98, 97, 95, 0],
}
df_s4 = pd.DataFrame(tabla_s4) #Convertimos el diccionario en un dataframe de pandas
print(df_s4)
```

```python
#Caso de la molécula benzoxazole, s5
print("Sustrato s5, benzoxazole")
display(Image(filename="s5_benzoxazole.png")) #Mostramos la estructura y posiciones potencialmente marcables con deuterio
tabla_s5= {
    "Catalizador": ["Ru·IMes", "Ru90Co10·IMes", "Ru50Co50·IMes", "Ru10Co90·IMes"],
    "%Red": [59, 30, 0, 0],
    "A": [96, 69, 14, 0],
    "B": [59, 8, 3, 0],
}
df_s5 = pd.DataFrame(tabla_s5) #Convertimos el diccionario en un dataframe de pandas
print(df_s5)
```

```python
#Queremos construir un heatmap, para facilitar la comprensión visual de estos valores numéricos. 
#Debemos cambiar la estructura de los df de pandas que hemos creado, de manera que toda la información de los cuatro catalizadores y cinco sustratos quede recogida de manera global.
#Vamos a crear una función para cambiar el formato al que queremos.

def convertir_a_formato_global(df, nombre_sustrato):
    posicion_columnas = [i for i in df.columns if i not in ("Catalizador", "%Red")]
    df_formato_global = pd.melt(df, id_vars=["Catalizador", "%Red"], value_vars=posicion_columnas,
                        var_name="Posición", value_name="%H/D")
    df_formato_global["Sustrato"] = nombre_sustrato
    return df_formato_global
```

```python
#Vamos a probar la función anterior para el sustrato s1
prueba = convertir_a_formato_global(df_s1, s1)
print(prueba)
print(prueba.shape)
```

Vale, tenemos una tabla para s1 donde se recoge todo el contexto de los resultados obtenidos con cada catalizador para ese sustrato. Resulta engorrosa de interpretar para el lector, en tanto que repite continuamente el %Red y el nombre del sustrato, pero nos va a ayudar a elaborar el heatmap. Usemos esa función para cada df.



```python
#Aplicamos a la función a cada df_sx y  sx
df_s1_formato_global = convertir_a_formato_global(df_s1, s1)
df_s2_formato_global = convertir_a_formato_global(df_s2, s2)
df_s3_formato_global = convertir_a_formato_global(df_s3, s3)
df_s4_formato_global = convertir_a_formato_global(df_s4, s4)
df_s5_formato_global = convertir_a_formato_global(df_s5, s5)

df_global = pd.concat([df_s1_formato_global, df_s2_formato_global, df_s3_formato_global, df_s4_formato_global, df_s5_formato_global], ignore_index=True)
#Dos cosas sobre la línea de arriba: con pd.concat, apilaremos los df_sx_formato_global cada uno debajo del anterior, 
#y con ignore_index=True forzamos a pandas a que genere un índice continuo en el conjunto global, 
#evitando así que se introduzcan índices confusos y/o repetitivos

#Por otro lado, esperamos que se genere un df uniendo las cinco tablas sin perder información, 
# por lo que esperamos que tenga 5 columnas (las que hemos introducido en la función convertir_a_formato_global)
# y un número de filas para cada sustrato igual al número de posiciones deuteradas x 4 (por el desglose de cuatro 
# catalizadores). Sustrato por sustrato, el número de filas será s1: (4x5), s2: (4x4), s3: (4x5), s4: (4x2), s5: (4x2), 72 filas en total. Comprobémoslo con print(df_global.shape) 
print(df_global.shape)

pd.set_option('display.max_rows', None) #Quiero ver el df global al completo, en cuyo caso voy a modificar brevemente la configuración  del máximo de filas de pandas
print(df_global) #Veamos el df_global que hemos creado
```

Ahora que ya hemos unido la información de manera global, vamos a pivotar la tabla. Además, vamos a resumir en una única celda mediante aggfunc="mean" el promedio de incorporación de deuterio entre todas las posiciones medidas. Es decir, buscamos una manera visual de comunicar cuánto deuterio incorpora cada catalizador en cada sustrato.

```python
df_pivot_exchange = df_global.pivot_table(index="Catalizador", columns="Sustrato", values="%H/D", aggfunc="mean")
#Con index="Catalizador" y columns="Sustrato" estamos diciendo qué información va en filas y qué información en columnas, y con values, el valor de incorporación de deuterio
#Por otro lado, queremos ordenar los catalizadores en orden decreciente de contenido en rutenio, así que:
orden_catalizadores = ["Ru·IMes", "Ru90Co10·IMes", "Ru50Co50·IMes", "Ru10Co90·IMes"]
orden_sustratos = ["2-phenylpyridine", "2,2′-bipyridyl", "2-methylquinoline", "imidazole", "benzoxazole"]
df_pivot_exchange = df_pivot_exchange.reindex(index=orden_catalizadores, columns=orden_sustratos)
print(df_pivot_exchange)
```

Vale, en la celda anterior, se nos genera un dataframe con la información que buscábamos, una relación compacta entre sustratos, catalizadores, y el promedio de incorporación de deuterio. Es interesante como primer acercamiento a la actividad de los nanocatalizadores. La tabla anterior responde a la pregunta ¿Cuánto intercambio isotópico hay, en conjunto, por catalizador y sustrato? Sin embargo nos genera un problema de índole química: se está omitiendo la columna de %Red. Debemos tener en cuenta que, aunque el proceso químico que nos interesa es el intercambio isotópico H/D, dicha reacción compite con la reacción de reducción. También necesitamos tener presente
esa información, para evaluar la actividad catalítica de nuestro nanomaterial por completo. Así, podemos responder a la pregunta ¿Cuánto compite el proceso no deseado, la reducción, con la actividad sí deseada (H/D)?

Construyamos un df para la reducción. 


```python
df_pivot_reduction = df_global.pivot_table(index="Catalizador", columns="Sustrato", values="%Red", aggfunc="mean")
df_pivot_reduction = df_pivot_reduction.reindex(index=orden_catalizadores, columns=orden_sustratos)
print(df_pivot_reduction)
```

```python
#Ahora, elaboremos heatmaps de seaborn para df_pivot_exchange y df_pivot_reduction
import seaborn as sns #Llamamos a seaborn, abreviamos como sns
import matplotlib.pyplot as plt #Llamamos a matplotlib, abreviamos como plt

#Para trabajar con los df anteriores, ciertos formalismos químicos los hemos ignorado deliberadamente, sobretodo de cara a facilitar la legibilidad del código. Pero para elaborar las gráficas, debemos utilizar la notación química correcta. Vamos a etiquetar los catalizadores
etiquetas_catalizadores = {
    "Ru·IMes": r"$\mathbf{Ru \cdot IMes}$",
    "Ru90Co10·IMes": r"$\mathbf{Ru_{90}Co_{10} \cdot IMes}$",
    "Ru50Co50·IMes": r"$\mathbf{Ru_{50}Co_{50} \cdot IMes}$",
    "Ru10Co90·IMes": r"$\mathbf{Ru_{10}Co_{90} \cdot IMes}$",
}
etiquetas_ordenadas = [etiquetas_catalizadores[c] for c in orden_catalizadores] #Relacionamos las etiquetas con el orden de catalizadores establecido varias celdas atrás.

fig, axes = plt.subplots(1, 2, figsize=(14, 5)) #Inicialmente creamos una figura con 1 fila y 2 columnas de huecos para gráficos. Nos devuelve fig, la figura completa y axes, una lista con los dos huecos individuales 

sns.heatmap(df_pivot_exchange, annot=True, fmt=".0f", cmap="crest", ax=axes[0], #Con annot=True le decimos a seaborn que muestre los valores numéricos, con fmt=".0f" quita los decimales, y además, que  dibuje en el hueco ax=axes[0]. Aparte, cmap="crest" es solo para el código de colores
            cbar_kws={'label': '% H/D (media)'}, yticklabels=etiquetas_ordenadas) #Asignamos los valores que va a tomar el heatmap, de intercambio H/D
axes[0].set_title("Intercambio promedio de H/D") #Titulamos individualmente el mapa de intercambio H/D

sns.heatmap(df_pivot_reduction, annot=True, fmt=".0f", cmap="Reds", ax=axes[1], #Le decimos a seaborn que dibuje en el hueco ax=axes[0]
            cbar_kws={'label': '% Reducción'}, yticklabels=etiquetas_ordenadas) #Asignamos los valores que va a tomar el heatmap, de reducción
axes[1].set_title("Reducción(%Red)") #Titulamos individualmente el mapa de reducción

plt.suptitle("Nanopartículas RuCo·IMes: actividad de intercambio H/D vs. reducción") #Titulamos el conjunto
plt.tight_layout()
plt.savefig("H_DDvsRed.png", dpi=300, bbox_inches="tight") #Guardamos los heatmaps imagen de manera local
plt.show()
```

Los dos mapas anteriores nos ayudan visualmente a correlacionar la composición del nanomaterial con el comportamiento catalítico del mismo, sustrato a sustrato. El comportamiento general es que, debido a la gran actividad catalítica que suelen mostrar los catalizadores basados en rutenio, la reacción evolucione hacia la reducción del sustrato; que en realidad es un proceso no deseado. Como estamos ante un catalizador bimetálico, se está añadiendo cobalto por dos motivos: El primero es para modular la actividad catalítica, disminuyendo la reducción, y facilitando que se de el proceso de intercambio isotópico (deseado). El segundo motivo es económico: el rutenio es un metal noble, caro y escaso, mientras que el cobalto es más abundante.

Cabe aclarar que, si bien el %Reducción es absoluto, esto es, va referido al número de moles totales de sustrato presentes inicialmente que experimenta reducción, el %intercambio H/D es un valor que surge al promediar la incorporación de deuterio en cada posición, sólo en moléculas que no la experimentan.  

Estos mapas muestran un cierto comportamiento en ese sentido. Pongamos el caso de la molécula de 2-phenylpyridine (primera columna). Con el catalizador **Ru·IMes**, la reducción del sustrato es de un 66%. Del 34% restante, el promedio de incorporación de deuterio en cada posición es de un 47%. Con el catalizador **$Ru_{90}$ $Co_{10}$·IMes**, la reducción del sustrato es de un 55%. Del 45% restante, el promedio de incorporación de deuterio en cada posición es de un 48%. Es decir, disminuir ligeramente la cantidad de rutenio y sustituirlo por cobalto ha tenido el efecto buscado, disminuir la reducción, manteniendo un nivel de incorporación de deuterio similar. Podríamos desglosar toda la relación catalizador a catalizador y sustrato a sustrato, pero visualmente se entiende mejor con heatmaps de seaborn. 

Observando los resultados de incorporación de H/D promedio, obtenemos una información muy curiosa. Para un catalizador dado, en general, disminuir la cantidad de rutenio en el nanomaterial, aumentando la de cobalto, tiene el efecto de disminuir tanto la reducción como la incorporación de deuterio promedio. Debería ser posible encontrar un punto óptimo, donde se maximice la incorporación de deuterio y se minimice la reducción. Sin embargo, al representar así los datos, se ha ocultado una información valisísima: la selectividad por las distintas posiciones "deuterables". En las celdas posteriores, vamos a recuperarla y representarla también. 

```python
#Creemos un df de pandas donde sólo se observe el porcentaje H/D incorporado absoluto, no promedio, desglosado posición a posición
df_s1_selectividad = df_global[df_global["Sustrato"] == s1].pivot_table( #Hacemos filtrado booleano, para devolver del df_global sólo lo referido al sustrato s1
    index="Catalizador", columns="Posición", values="%H/D"
)
df_s1_selectividad = df_s1_selectividad.reindex(index=orden_catalizadores)
print(df_s1_selectividad)
```

Perfecto, observamos fácilmente sólo los datos que queremos, desglosados por posición.  En el siguiente paso vamos a mostrar heatmaps de seaborn desglosados por posición, que le servirán al lector para detectar visualmente la selectividad de cada catalizador.   

```python

sustratos_lista = [s1, s2, s3, s4, s5] #Asociamos los nombres de los sustratos, previamente asignamos a las variables s1-s5, a una lista

fig, axes = plt.subplots(2, 3, figsize=(20, 10)) #Esta línea hace que axes no sea una lista consecutiva simple, sino un grid 2 x 3, para que aparezcan los heatmaps de selectividad en dos alturas. Sin embargo, cuidado, porque solo tenemos 5 entradas, pero se ha generado una matriz de ejes donde debería hacer 6 (2x3), debemos "eliminarla" luego. Con figsize=(20, 10) condicionamos el tamaño de ancho y alto
axes = axes.flatten() #Convertimos la matriz 2 x 3 en una lista plana con un orden de lectura dado, y así poder utilizar cómodamente zip() y enumerate()

for i, (ax, sustrato) in enumerate(zip(axes, sustratos_lista)): #Con este bucle for ax recorremos dos listas a la vez: en un eje, cada hueco del panel, y por otro lado, la lista de sustratos, sustratos_lista
    df_sub_pivot = df_global[df_global["Sustrato"] == sustrato].pivot_table( #Como antes, hacemos un filtrado booleano, para en cada caso devolver datos referidos sólo a ese sustrato
        index="Catalizador", columns="Posición", values="%H/D"
    )
    df_sub_pivot = df_sub_pivot.reindex(orden_catalizadores)
    mostrar_etiquetas = (i == 0 or i == 3)  # Etiquetamos el primer panel de cada fila de heatmaps con los nombres de los catalizadores, para no saturar al lector. En cualquier caso, los catalizadores son los mismos
    sns.heatmap(df_sub_pivot, annot=True, fmt=".0f", cmap="GnBu",
                vmin=0, vmax=100, ax=ax, cbar=False, #Con vmin=0, vmax=100 fijamos la escala de colores para los cinco heatmaps a la vez, para que las diferencias sean más apreciables. De otra forma, el valor máximo de un heatmap (pongamos 17%) tendría el mismo color que en otro (pongamos 80%), lo cual queremos evitar
               annot_kws={"size": 13}, linewidths=0.5, linecolor="white",  #annot_kws={"size": XX} controla específicamente el tamaño de los números dentro de las celdas; linewidths=0.5, linecolor="white", pone una línea blanca de un cierto grosor, para facilitar distinguirlas al ojo 
    yticklabels=etiquetas_ordenadas if mostrar_etiquetas else False)
    ax.set_title(sustrato, fontsize=13)
    ax.tick_params(axis='both', labelsize=12) #Controla las etiquetas de los ejes de cada panel individual, incluyendo nombres de posiciones y panel de catalizadores

axes[-1].axis('off')  # "eliminamos" el sexto hueco: solo tenemos 5 sustratos, no 6    

plt.suptitle("Selectividad de incorporación H/D por posición, catalizador y sustrato", fontsize=15)
plt.tight_layout()
plt.savefig("Heatmaps_Selectividad.png", dpi=300, bbox_inches="tight") #Guardamos los heatmaps imagen de manera local
plt.show()
```

Al desglosar la incorporación de deuterio por posición, puede observarse una tendencia que antes había pasado despercibida. Si bien disminuir la cantidad de rutenio en el nanomaterial, aumentando la de cobalto, tiene el efecto de disminuir tanto la reducción como la incorporación de deuterio promedio; en ciertos sustratos (2-phenylpyridine, 2-methylquinoline, imidazole) se mejora mucho la selectividad hacia ciertas posiciones al utilizar catalizadores algo más ricos en cobalto, como **$Ru_{50}$ $Co_{50}$·IMes**. En ciertos casos, se observa que ciertas posiciones son deuteradas en un 90% o superior con dicho catalizador. En otras palabras: aunque en global no sea tan activo, **$Ru_{50}$ $Co_{50}$·IMes** permite deuterar selectivamente ciertas posiciones de algunos sustratos (A en 2-phenylpyridine, A y E en 2-methylquinoline) o, en otros casos, deuterar casi por completo todas las posiciones posibles (imidazole). Aparentemente, el punto óptimo (mínima reducción, máxima incorporación de deuterio), debe encontrarse en torno a los catalizadores centrales, **$Ru_{50}$ $Co_{50}$·IMes** o **$Ru_{90}$ $Co_{10}$·IMes**, ya que los catalizadores de los extremos se orientan demasiado hacia la reducción (**Ru·IMes**)
o no muestran actividad (**$Ru_{10}$ $Co_{90}$·IMes**)

Una vez visualizados los datos, vamos a utilizar scikit-learn para llevar a cabo una regresión. Comencemos preparando los datos. Vamos a estudiar la correlación entre el aumento de la cantidad de cobalto en el catalizador y el comportamiento catalítico. En la Tabla 1 del paper de donde se han tomado los datos, los autores dan valores reales de %Co y %Ru medidos experimentalmente mediante análisis por ICP. Debemos utilizar esos datos reales, y asociarlos a las etiquetas empleadas. el catalizador **$Ru_{90}$ $Co_{10}$·IMes** no tiene una composición centeisimal de un 90%wt de rutenio y un 10%wt de cobalto, sino que el ligando orgánico empleado tambien es un componente importante del mismo. Nuestro modelo de sci-kit debe ser consciente de ello. Empecemos con los datos de reducción. Vamos a realizar la regresión únicamente con los datos de %Co (a modo de predictor de la reducción promedia), por dos motivos:

1) %Co y %Ru están correlacionados inversamente, aunque no son % complementarios debido a la cantidad de ligando orgánico. Si usáramos %Co y %Ru como predictores en una regresión con tan pocos datos, sólo 4 catalizadores, estaríamos sobreajustando el modelo estadístico, tendríamos 4 puntos y 2 "predictores". Para empezar, ya es problemático tener tan pocos puntos, como para además, tener dos variables... que encima, estarían cuasi inversamente relacionadas. Dejemos %Ru a un lado, no aporta demasiada información respecto a los problemas que causa tenerlo en cuenta.
2) Los autores del paper cuyos datos estamos utilizando, señalan que %Co es la variable que controla la tendencia en mayor medida. Centrémonos en el sentido químico de los experimentos. 

 En las próximas celdas, vamos a construir un df con el %Co y el %Red promedio para cada catalizador, teniendo en cuenta los cinco sustratos. Se pretende correlacionar los datos mediante un pequeño modelo. Se recomienda ver el subset **Reducción(%Red)** del gráfico **Nanopartículas RuCo·IMes: actividad de intercambio H/D vs. reducción** (celda 15).

```python
porcentaje_co = { #Creamos un diccionario con los datos de porcentaje de cobalto en masa, utilizando los datos reales de la tabla "Table 1" del paper, Dalton Trans. (2025) 54 (46): 17250–17258 
    "Ru·IMes": 0,
    "Ru90Co10·IMes": 2,
    "Ru50Co50·IMes": 14,
    "Ru10Co90·IMes": 19,
}

df_red_Co_wt = pd.DataFrame({ #Construimos un dataframe de pandas con los porcentajes de cada metal para cada catalizador y la red promedio obtenida
    "Catalizador": orden_catalizadores,
    "%Co": [porcentaje_co[c] for c in orden_catalizadores],
    "%Red_medio": df_pivot_reduction.mean(axis=1), #Promediamos el porcentaje de reducción obtenido con cada sustrato 
})
df_red_Co_wt_bin = df_red_Co_wt.drop('Catalizador', axis=1)
print(df_red_Co_wt_bin)
```

```python
from sklearn.linear_model import LinearRegression #Importamos LinearRegression

X = df_red_Co_wt_bin[["%Co"]]   # Ponemos doble corchete porque sklearn necesita un dataframe de una sola columna. 
y = df_red_Co_wt_bin["%Red_medio"]

modelo = LinearRegression() 
modelo.fit(X, y) #Ajustamos la recta que mejor se aproxima a los 4 puntos

print("Pendiente: ", modelo.coef_[0])
print("Ordenada en el origen (Reducción teórica con 0% Co):", modelo.intercept_)
print("R²:", modelo.score(X, y))
```

```python
import numpy as np #Importamos numpy

X = df_red_Co_wt_bin[["%Co"]]   # Ponemos doble corchete porque sklearn necesita un dataframe de una sola columna. 
y = df_red_Co_wt_bin["%Red_medio"]
plt.figure(figsize=(7, 5))
plt.scatter(X, y, color='steelblue', s=100, label='Datos experimentales')

x_recta = pd.DataFrame(np.linspace(0, 20, 100), columns=["%Co"])  # <-- Le ponemos al dataframe el nombre de columna "%Co"
y_recta = modelo.predict(x_recta) #Usamos la recta ya ajustada para calcular de acuerdo a este modelo el %Red que le correspondería a cada valor de %Co generado en la línea anterior
plt.plot(x_recta, y_recta, color='firebrick', linestyle='--', label=f'Regresión (R²={modelo.score(X,y):.2f})') #Formateamos el valor de R^2 para mostrar sólo 2 decimales

plt.xlabel('% Co (peso, ICP)')
plt.ylabel('% Reducción promedio')
plt.title('Correlación entre contenido en cobalto de los catalizadores y tendencia a producir reducción del sustrato')
plt.legend()
plt.savefig("CorrelCoRed.png", dpi=300, bbox_inches="tight") #Guardamos de manera local
plt.show()
```

En términos químicos, la gráfica anterior (Correlación entre contenido en cobalto de los catalizadores y tendencia a producir reducción del sustrato) muestra un decaimiento lineal en lo que respecta a la tendencia del catalizador a disminuir su actividad hacia la reducción; según aumenta su contenido en cobalto. Sin embargo, se debe ser cauto con dicha observación por dos motivos principales. En primer lugar, el número total de reacciones analizadas es muy bajo, un total de 20 catálisis (5 sustratos x 4 catalizadores) y el rango de %Co del que se dispone es estrecho, con catalizadores que muestran entre un 0% y un 20% en peso de cobalto. 

Ya hemos correlacionado el contenido en cobalto de los catalizadores y la tendencia a producir reducción del sustrato (reacción no deseada). A continuación, vamos a correlacionar el contenido en cobalto conla tendencia a producir intercambio isotópico (reacción sí deseada). En la siguiente celda, vamos a construir un df con el %Co y el %H/D promedio para cada catalizador, teniendo en cuenta los cinco sustratos. Se recomienda ver el subset **Intercambio promedio de H/D** del gráfico **Nanopartículas RuCo·IMes: actividad de intercambio H/D vs. reducción** (celda 15).

```python
df_HD_Co_wt = pd.DataFrame({
    "Catalizador": orden_catalizadores,
    "%Co": [porcentaje_co[c] for c in orden_catalizadores],
    "%H/D": df_pivot_exchange.mean(axis=1),
})
df_HD_Co_wt_bin = df_HD_Co_wt.drop('Catalizador', axis=1)
print(df_HD_Co_wt_bin)
```

Muy bien, teniendo ya este dataframe, vamos a construir un nuevo modelo, con los datos de %H/D. Aparentemente también hay un decaimiento lineal. 

```python
X_HD = df_HD_Co_wt_bin[["%Co"]] #Usamos la variable X_HD por precaución y no utilizar de nuevo X, usada anteriormente
y_HD = df_HD_Co_wt_bin["%H/D"]  #Usamos la variable Y_HD, ídem a lo anterior

modelo_HD = LinearRegression()
modelo_HD.fit(X_HD, y_HD)

print("Pendiente:", modelo_HD.coef_[0])
print("Ordenada en el origen: (Incorporación teórica con 0% Co)", modelo_HD.intercept_)
print("R²:", modelo_HD.score(X_HD, y_HD))
```

El valor de $R^{2}$=0.91 en esta segunda regresión no es muy alto. En conjunto, tenemos dos regresiones con la misma variable (%Co) pero con una calidad de ajuste lineal distinta (0.91 vs 0.97). Parece que la presencia de cobalto en el catalizador "apaga" la reducción de manera muy predecible, pero su efecto sobre la incorporación de deuterio es algo menos lineal. No es que el cobalto esté simplemente diluyendo al rutenio, por lo que baja su actividad reductora, es que en sí mismo la presencia de cobalto modula la incorporación de deuterio en el sustrato. Veamos visualmente ambas correlaciones: 

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

x_recta = pd.DataFrame(np.linspace(0, 20, 100), columns=["%Co"])

# Gráfica izquierdo: %Red vs %Co
axes[0].scatter(X, y, color='steelblue', s=100, label='Datos experimentales')
y_recta_red = modelo.predict(x_recta)
axes[0].plot(x_recta, y_recta_red, color='firebrick', linestyle='--',
             label=f'Regresión (R²={modelo.score(X, y):.2f})')
axes[0].set_xlabel('% Co (peso, ICP)')
axes[0].set_ylabel('% Reducción promedio')
axes[0].set_title('%Reducción promedio vs. % Co, proceso químico no deseado')
axes[0].legend()

# Panel derecho: %H/D vs %Co
axes[1].scatter(X_HD, y_HD, color='seagreen', s=100, label='Datos experimentales')
y_recta_HD = modelo_HD.predict(x_recta)
axes[1].plot(x_recta, y_recta_HD, color='darkorange', linestyle='--',
             label=f'Regresión (R²={modelo_HD.score(X_HD, y_HD):.2f})')
axes[1].set_xlabel('% Co (peso, ICP)')
axes[1].set_ylabel('% H/D promedio')
axes[1].set_title('Intercambio H/D vs. % Co, proceso químico deseado')
axes[1].legend()

plt.suptitle('Efecto del contenido de Co en el catalizador: reacción deseada vs. reacción competitiva')
plt.tight_layout()
plt.savefig("comparacion_regresiones_Co.png", dpi=300, bbox_inches="tight")
plt.show()
```

De las gráficas anteriores, puede observarse que en ambos casos la pendiente es negativa. Los dos procesos químicos se ven menos favorecidos en global conforme aumenta la cantidad de cobalto (pendientes de -2.12 en reducción y -3.32 en H/D). No se pueden forzar los datos: no existe un óptimo que alcanzar diseñando nuevos catalizadores, en lo que respecta a minimizar %Red y maximizar %H/D. Pero sí podemos calcular una relación de "preferencia" por un tipo de proceso u otro. La relación (%H/D)/(%Red) sí que puede tener un valor máximo para un catalizador. 

```python
df_comparativa_HD_red = pd.DataFrame({
    "Catalizador": orden_catalizadores,
    "%Co": [porcentaje_co[c] for c in orden_catalizadores],
    "%H/D_medio": df_pivot_exchange.mean(axis=1).values,
    "%Red_medio": df_pivot_reduction.mean(axis=1).values,
})
df_comparativa_HD_red["Cociente (H/D:Red)"] = df_comparativa_HD_red["%H/D_medio"] / df_comparativa_HD_red["%Red_medio"]
print(df_comparativa_HD_red)
```

Observando la tabla anterior, es fácil observar que el catalizador que ofrece mejor relación entre el proceso deseado y el indeseado es **$Ru_{50}$ $Co_{50}$ ·IMes** (Cociente (H/D:Red = 2.049351)). Si bien ambos procesos decaen al disminuir la cantidad de rutenio en el catalizador, el catalizador mencionado es el que ofrece el mejor compromiso relativo entre ambos procesos químicos competitivos. Cabe destacar que para el último catalizador, **$Ru_{10}$ $Co_{90}$ ·IMes** no tiene sentido matemático definir dicho cociente, debido a la baja actividad presentada para ambos procesos. Estos datos son consistentes con la visualización que obtuvimos en la gráfica **Selectividad de incorporación H/D por posición, catalizador y sustrato**, de la celda 19.
